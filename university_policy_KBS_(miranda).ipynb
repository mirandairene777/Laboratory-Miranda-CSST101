1. Debate Arguments

```r
argument <- data.frame(
  Argument_ID = c("A001","A002","A003","A004","A005","A006"),
  Position = c("For","For","For","Against","Against","Against"),
  Topic = c("Learning","Productivity","AI Literacy",
            "Academic Integrity","Dependency","Assessment"),
  Claim = c(
    "AI can provide students with immediate learning support.",
    "AI can help student complete routine academic tasks more efficiently.",
    "Allowing AI use can helps students develop AI literacy.",
    "Undisclosed AI-generated work may create academic integrity concerns.",
    "Excessive AI use may reduce students' independent problem-solving practice.",
    "AI use may make it more difficult to determine whether submitted work reflects student learning."
  )
)

argument
```

2. Evidence

```r
evidence <- data.frame(
  Evidence_ID = c("E001", "E002", "E003", "E004", "E005", "E006"),
  Argument_ID = c("A001", "A002", "A003", "A004", "A005", "A006"),
  Evidence_Type = c("Research", "Research", "Policy", "Policy", "Research", "Research"),
  Strength = c("Strong", "Moderate", "Strong", "Strong", "Moderate", "Strong")
)

evidence
```

3. Retrive arguments supporting the policy

```r
for_argument <- subset(
  argument,
  Position == "For"
)

print (for_argument)
```

4. Retrive arguments against the policy


```r
against_argument <- subset(
  argument,
  Position == "Against"
)

print (against_argument)
```

5. Connect arguments and evidence

```r
knowledge_base <- merge(
  argument,
  evidence,
  by = "Argument_ID"
)
print(knowledge_base)
```

6. Convert evidence strength into a score

```r
get_score <- function(strength){

  if (strength == "Strong"){
    return(3)
  }else if(strength == "Moderate"){
    return (2)
  }else if(strength == "Weak"){
    return(1)
  }else{
    return(0)
  }
}

get_score("Strong")
```

7. Add evidence scores

```r
knowledge_base$Evidence_Score <- sapply(
  knowledge_base$Strength,
  get_score
)
print(knowledge_base)
```

8. Classify evidence support

```r
classify_support <- function(score){

  if (score == 3){
    return("Strongly Supported")
  } else if (score == 2){
    return("Moderately Supported")
  } else if (score == 1){
    return("Weakly Supported")
  } else{
    return("No Support")
  }
}

classify_support(2)
```

9. Search by topic

```r
search_topic<- function(topic){
  result <- subset(
  knowledge_base,
  Topic == topic
  )

  if(nrow(result) == 0)(
    cat("Topic not found.\n")
  )else(
    print(result)
  )
}
```

```r
search_topic("Learning")
```
