1. Create a student database

```r
students <- data.frame(
STUDENT_ID = c("S001", "S002", "S003", "S004", "S005", "S006", "S007", "S008"),
Programming_grade = c(88,72,91,78,65,85,74,82),
Math_GRADE = c(85,70,88,80,68,90,73,79),
Attendance = c(92,80,95,88,70,91,76,85),
Prerequisite = c("Yes", "No", "Yes", "Yes", "No", "Yes", "No", "Yes")
)
students
```

6. Extract student information

```r
programming <- student$Programming_grade
attendance <- student$Attendance
mathGrade <- student$Math_GRADE
studentID <- student$STUDENT_ID
Prerequisite <- student$Prerequisite

cat("\nStudent ID        : ", studentID)
cat("\nProgramming Grade : ", programming)
cat("\nAttendace         : ", attendance)
cat("\nMath Grade        : ", mathGrade)
cat("\nPrerequisite      : ", Prerequisite)
```

7. Implement the rules

```r
classify_student <- function
(programming_grade, math_grade, attendance_percentage, has_prerequisite)
{
  if (programming_grade >= 80 &&
    math_grade >= 75 &&
    attendance_percentage >= 80 &&
    has_prerequisite == "Yes")
    {
    return("Eligible")
  } else if (programming_grade >= 75 &&
    math_grade >= 70 &&
    attendance_percentage >= 75)
    {
    return("Conditionally Eligible")
  } else {
    return("Not Eligible")
  }
}

student_classification <- classify_student(programming, mathGrade, attendance, Prerequisite)
cat("Student ID:", studentID, "is", student_classification, "\n")
```

8. Apply all functions to all students

```r
students$Eligibility <- mapply(classify_student,
                                   programming_grade = students$Programming_grade,
                                   math_grade = students$Math_GRADE,
                                   attendance_percentage = students$Attendance,
                                   has_prerequisite = students$Prerequisite)

print(students)
```

9. create recommendation rules

```r
get_recommendation <- function(eligibility) {

  if (eligibility == "Eligible") {
    return("Student may Enroll")

  } else if (eligibility == "Conditionally Eligible") { # Fixed typo here
    return("Consult Academic Adviser")

  } else {
    return("Complete requirements before enrollment")
  }
}

students$Recommendation <- sapply(
  students$Eligibility,
  get_recommendation
)

students
```
